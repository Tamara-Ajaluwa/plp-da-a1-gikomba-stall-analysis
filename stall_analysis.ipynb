```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,

               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2



# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {

    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},

    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},

    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},

    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},

    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}
```

```python
#Task 1


# Code for total and average sales
total_sales = sum(daily_sales)
avg_sales = total_sales / len(daily_sales)

# Code for highest and lowest daily sales
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Code to identify index positions of best and worst sales days
best_day_idx = daily_sales.index(highest_sales)
worst_day_idx = daily_sales.index(lowest_sales)

# Code to get corresponding day name and determine week number (positions 0-6 = Week 1, 7-13 = Week 2)
best_day_name = days[best_day_idx]
best_week = 1 if best_day_idx < 7 else 2

worst_day_name = days[worst_day_idx]
worst_week = 1 if worst_day_idx < 7 else 2

# Code to print the results
print(f"Total Sales: KES {total_sales:,.2f}")
print(f"Average Daily Sales: KES {avg_sales:,.2f}")
print(f"Highest Sales: KES {highest_sales:,.2f} ({best_day_name}, Week {best_week})")
print(f"Lowest Sales: KES {lowest_sales:,.2f} ({worst_day_name}, Week {worst_week})")
```

```python
# Task 2

def week_totals(sales):
    """Code for calculating and returning total sales for Week 1 and Week 2 as a tuple using list slicing."""
    week1_total = sum(sales[:7])
    week2_total = sum(sales[7:])
    return (week1_total, week2_total)

# Code to call function and unpack two values
week1_total, week2_total = week_totals(daily_sales)

# Code to calculate week-on-week growth percentage
wow_growth = ((week2_total - week1_total) / week1_total) * 100

# Code to print the results
print(f"Week 1 Total Sales: KES {week1_total:,.2f}")
print(f"Week 2 Total Sales: KES {week2_total:,.2f}")
print(f"Week-on-Week Growth: {wow_growth:.2f}%")
```

```python
# Task 3

def classify_day(amount, average):
    """Code to classify the sales amount as Strong, Normal, or Weak relative to the average."""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"

# Code to count frequency of each classification using dictionary counting pattern
category_counts = {"Strong": 0, "Normal": 0, "Weak": 0}

for sale in daily_sales:
    category = classify_day(sale, avg_sales)
    category_counts[category] += 1

# Code to print the results
print("Day Classification Breakdown:")
for cat, count in category_counts.items():
    print(f"  {cat}: {count} day(s)")
```

```python
# Task 4

# Code to separate sales into weekdays (Mon-Fri) and weekends (Sat-Sun)
weekdays = [daily_sales[i] for i in range(14) if i % 7 < 5]
weekends = [daily_sales[i] for i in range(14) if i % 7 >= 5]

avg_weekday = sum(weekdays) / len(weekdays)
avg_weekend = sum(weekends) / len(weekends)

# Code to calculate percentage difference of weekend sales compared to weekday sales
pct_diff = ((avg_weekend - avg_weekday) / avg_weekday) * 100

# Code to print the results
print(f"Average Weekday Sales: KES {avg_weekday:,.2f}")
print(f"Average Weekend Sales: KES {avg_weekend:,.2f}")
print(f"Weekend vs Weekday Difference: {pct_diff:.2f}%")
```

```python
# Task 5

def check_restock(stock_dict):
    """
    Identifies products at or below reorder level and calculates gross margin percentage.
    Returns a list of tuples: (product_name, current_qty, gross_margin_pct).
    """
    restock_list = []
    for item, data in stock_dict.items():
        if data["qty"] <= data["reorder"]:
            margin = ((data["price"] - data["cost"]) / data["price"]) * 100
            restock_list.append((item, data["qty"], margin))
    return restock_list

# Xode to execute function using the stock dictionary
items_to_reorder = check_restock(stock)

# Code to print the results
print("Items Needing Restock:")
for item, qty, margin in items_to_reorder:
    print(f"  - {item}: Current Qty = {qty}, Gross Margin = {margin:.2f}%")
```

```python
# Task 6

report = f"""
========================================
       STALL PERFORMANCE REPORT
========================================
Total Sales:          KES {total_sales:,.2f}
Average Daily Sales:  KES {avg_sales:,.2f}
Week 1 Total Sales:   KES {week1_total:,.2f}
Week 2 Total Sales:   KES {week2_total:,.2f}
Week-on-Week Growth:  {wow_growth:.2f}%
Average Weekday Sales:KES {avg_weekday:,.2f}
Average Weekend Sales:KES {avg_weekend:,.2f}
Weekend vs Weekday:   {pct_diff:.2f}%
========================================
"""
print(report)
```
